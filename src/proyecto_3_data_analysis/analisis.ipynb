```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn
```

```python
df = pd.read_csv("/Users/rodrigomayagonzalez/Documents/Python Proyectos aprender/Proyecto_3_Data_Analysis/archive/DataCoSupplyChainDataset.csv",encoding='latin-1')
df
```

```python
data_limpia = df[['Type', 'Days for shipping (real)', 'Days for shipment (scheduled)', 'Delivery Status', 'Late_delivery_risk',
                  'Category Name', 'Customer City', 'Customer Country', 'Order Country', 'order date (DateOrders)',
                  'Order Id', 'Sales per customer', 'Order Item Product Price', 'Order Item Quantity', 'Order Item Total',
                  'Order Status', 'shipping date (DateOrders)', 'Shipping Mode']]
data_limpia.columns
```

```python
data_limpia
```

```python
data_limpia.info()
```

```python
data_limpia['order date (DateOrders)'] = pd.to_datetime(data_limpia['order date (DateOrders)'])
data_limpia['shipping date (DateOrders)'] = pd.to_datetime(data_limpia['shipping date (DateOrders)'])
```

```python
data_limpia.info()
```

```python
data_limpia['Mes'] = data_limpia['shipping date (DateOrders)'].dt.month_name()
orden_meses = [
    'January', 'February', 'March', 'April', 'May', 'June', 
    'July', 'August', 'September', 'October', 'November', 'December'
]
data_limpia['Mes'] = pd.Categorical(data_limpia['Mes'], categories=orden_meses, ordered=True)
data_limpia['Year'] = data_limpia['shipping date (DateOrders)'].dt.year
data_limpia
```

```python
data_pais_group = data_limpia.groupby('Order Country')['Order Id'].nunique().reset_index()
total = data_pais_group['Order Id'].sum()
data_pais_group['Porcentaje'] = round(data_pais_group['Order Id'] / total *100,2)
data_pais_group = data_pais_group.sort_values(by='Porcentaje', ascending=False)
data_pais_group = data_pais_group.head(20)
```

```python
fig,ax = plt.subplots(figsize = (5,4))
barras = ax.barh(data_pais_group['Order Country'], data_pais_group['Porcentaje'])

for barra in barras:
    ancho = barra.get_width()
    ax.text(ancho + 1, barra.get_y() + barra.get_height()/2, f'{ancho}%',
            va='center', ha='left', fontsize=8)

ax.set_title('Top 20 Paises con mayor porcentaje de ordenes')
ax.set_xlim(0,45)

plt.show()
```

Pero estados unidos tiene la mayor cantidad de ventas? y como esta distribuido ese porcentaje por mes?

```python
piv = data_limpia.pivot_table(index='Order Country', columns=['Year','Mes'], values='Order Id', aggfunc='nunique', fill_value=0, observed= True)
piv['total'] = piv.sum(axis=1)
plt.figure(figsize=(20,10))
seaborn.heatmap(piv, cmap='inferno', center=1, cbar_kws={'label': 'vs promedio del país'})
```

```python
piv.sort_values(by='total', ascending=False)
```

```python
data_ingresos_grouped = data_limpia.groupby('Order Country')['Order Item Total'].sum().reset_index()
data_ingresos_grouped = data_ingresos_grouped.sort_values(by= 'Order Item Total', ascending=False)
data_ingresos_grouped
```

```python
data_delivery_status = data_limpia.groupby('Delivery Status')['Order Id'].nunique().reset_index()
total_ds = data_delivery_status['Order Id'].sum()
data_delivery_status['Porcentaje'] = round(data_delivery_status['Order Id'] / total_ds *100, 2)
data_delivery_status = data_delivery_status.sort_values(by='Porcentaje', ascending=False)
data_delivery_status
```
